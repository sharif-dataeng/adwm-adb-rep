Schema registry utilities for Adventure Works ingestion.

Notebook contents:

* Define source schemas by source domain
* Assemble the `schema_registry` dictionary used by bronze and silver processing notebooks

Usage notes:

* Keys in `schema_registry` are lowercase `<schema>.<table>` identifiers
* The registry is used to enforce expected schemas during ingestion and downstream processing
* Update this notebook when a new source table is added or an existing source schema changes
