This notebook creates the reusable gold-layer territory BI view `adwm_wh.gold.vw_bi_sales_by_territory` for downstream regional reporting.

Scope:

* Create or replace the territory BI view in `adwm_wh.gold`
* Reuse `adwm_wh.gold.vw_bi_factsales_base` as the source layer
* Summarize sales performance by territory and country
* Run lightweight validation for top territory results

Business grain:

* One row per territory and country

```python
%sql
CREATE OR REPLACE VIEW adwm_wh.gold.vw_bi_sales_by_territory AS
SELECT
  TerritoryName,
  CountryRegion,
  COUNT(*) AS SalesLineCount,
  COUNT(DISTINCT SalesOrderNumber) AS SalesOrderCount,
  COUNT(DISTINCT CustomerKey) AS CustomerCount,
  SUM(OrderQuantity) AS TotalOrderQuantity,
  CAST(SUM(SalesAmount) AS DECIMAL(19,4)) AS TotalSalesAmount,
  CAST(SUM(TotalCost) AS DECIMAL(19,4)) AS TotalCost,
  CAST(SUM(DiscountAmount) AS DECIMAL(19,4)) AS TotalDiscountAmount,
  CAST(SUM(GrossMargin) AS DECIMAL(19,4)) AS GrossMargin
FROM adwm_wh.gold.vw_bi_factsales_base
GROUP BY TerritoryName, CountryRegion;
```

```python
%sql
SELECT
  TerritoryName,
  CountryRegion,
  SalesLineCount,
  SalesOrderCount,
  CustomerCount,
  TotalOrderQuantity,
  CAST(TotalSalesAmount AS DECIMAL(19,2)) AS TotalSalesAmount,
  CAST(GrossMargin AS DECIMAL(19,2)) AS GrossMargin,
  CAST(TotalDiscountAmount AS DECIMAL(19,2)) AS TotalDiscountAmount
FROM adwm_wh.gold.vw_bi_sales_by_territory
ORDER BY TotalSalesAmount DESC, TerritoryName, CountryRegion
LIMIT 10;
```
