```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, BooleanType, TimestampType, FloatType, DoubleType, DateType
```

```python
employee_schema = StructType([
StructField('BusinessEntityID', IntegerType(), True),
StructField('NationalIDNumber', IntegerType(), True),
StructField('LoginID', StringType(), True),         
StructField('OrganizationNode', StringType(), True),
StructField('OrganizationLevel', StringType(), True),
StructField('JobTitle', StringType(), True),         
StructField('BirthDate', DateType(), True),       
StructField('MaritalStatus', StringType(), True),    
StructField('Gender', StringType(), True),           
StructField('HireDate', DateType(), True),         
StructField('SalariedFlag', BooleanType(), True),   
StructField('VacationHours', IntegerType(), True),   
StructField('SickLeaveHours', IntegerType(), True),  
StructField('CurrentFlag', BooleanType(), True),      
StructField('rowguid', StringType(), True),         
StructField('ModifiedDate', DateType(), True)    
    ])\
```

```python
df = spark.read.format('csv')\
    .option("header", "true")\
    .schema(employee_schema)\
    .load('/Volumes/adwm-dev-catalog/sharif_vol/sharif_sot_v1/AdventureWorkMock/HumanResources/incremental/Employee/')
```
