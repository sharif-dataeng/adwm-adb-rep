Unity Catalog external location setup for the Adventure Works warehouse.

Notebook contents:

* Create input widgets for Azure storage details
* Create a Unity Catalog storage credential reference
* Create the external location for the catalog managed storage path
* Create the external location for the landing files volume path

Prerequisites:

* A usable Unity Catalog storage credential must already exist
* Azure storage permissions must allow Unity Catalog access to the target containers
* Update the storage credential name in this notebook before running if needed

Expected result:

* A Unity Catalog external location exists for the catalog managed storage path
* A Unity Catalog external location exists for the landing files volume path

```python
# Create input widgets for storage account, containers, and storage credential
dbutils.widgets.text('storage_account', 'adwgen2')
dbutils.widgets.text('catalog_container', 'uc-adwm')
dbutils.widgets.text('volume_container', 'sharif')
dbutils.widgets.text('storage_credential', 'ups_ws')

# Use widget values in variables
storage_account = dbutils.widgets.get('storage_account')
container = dbutils.widgets.get('catalog_container')
volume_container = dbutils.widgets.get('volume_container')
storage_credential = dbutils.widgets.get('storage_credential')

catalog_external_location_name = 'adwm_wh_managed_loc'
catalog_external_location_url = f"abfss://{container}@{storage_account}.dfs.core.windows.net/adwm_wh"
volume_external_location_name = 'adwm_wh_landing_loc'
volume_external_location_url = f"abfss://{volume_container}@{storage_account}.dfs.core.windows.net/"
```

```python
catalog_external_location_sql = f"""
CREATE EXTERNAL LOCATION IF NOT EXISTS `{catalog_external_location_name}`
URL '{catalog_external_location_url}'
WITH (STORAGE CREDENTIAL `{storage_credential}`)
COMMENT 'External location for Adventure Works catalog managed storage'
"""

volume_external_location_sql = f"""
CREATE EXTERNAL LOCATION IF NOT EXISTS `{volume_external_location_name}`
URL '{volume_external_location_url}'
WITH (STORAGE CREDENTIAL `{storage_credential}`)
COMMENT 'External location for Adventure Works landing files volume'
"""

spark.sql(catalog_external_location_sql)
spark.sql(volume_external_location_sql)
```
